```python
import pandas as pd
import numpy as np


CONST = 28.8    
FACTOR = 0.6022 


df = pd.read_excel("Plasmon Energy-Layered TMHs-input.xlsx")


df["Vm (cm3/mol)"] = (df["Vcell (Å^3)"] * FACTOR) / df["cell_formula_units_Z"]


df["ħω_p (eV)"] = CONST * np.sqrt(df["valence_sp_per_fu"] / df["Vm (cm3/mol)"])



df.to_excel("Plasmon Energy-Layered TMHs-output.xlsx", index=False)


print(df[["Material", "Vm (cm3/mol)", "ħω_p (eV)"]])
```
