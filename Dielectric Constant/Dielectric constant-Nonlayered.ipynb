```python
import pandas as pd
import numpy as np


input_file = "Dielectric constant-Non Layered TMHs-input.xlsx"
output_file = "Dielectric constant-Non Layered TMHs-output.xlsx"


df = pd.read_excel(input_file)


required_cols = ["Material", "Eg (eV)", "ħω_p (eV)", "Ef (eV)"]
for col in required_cols:
    if col not in df.columns:
        raise ValueError(f"Missing required column: {col}")

Eg = df["Eg (eV)"].astype(float)
hbar_omega_p = df["ħω_p (eV)"].astype(float)
Ef = df["Ef (eV)"].astype(float)


x = Eg / (4.0 * Ef)
bracket = (1.0 - x + (1.0 / 3.0) * (x ** 2))


hbar_omega_g = np.sqrt(hbar_omega_p**2 + Eg**2)
eps_infinity = 1.0 + (hbar_omega_g / Eg) ** 2 * bracket


out_df = pd.DataFrame({
    "Material": df["Material"],
    "ε∞": eps_infinity,
})

out_df.to_excel(output_file, index=False)
print(f"✅ Done! Results saved as '{output_file}'")
```
